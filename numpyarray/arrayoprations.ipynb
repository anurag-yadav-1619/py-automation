```python
for i in range(0,9,3):
   for j in range(0,9,3):
      n=s[i:i+3,j:j+3]
      print(n.sum())
        
```

```python
s.sum()
```

```python
b=np.sum(s,axis=1)
for i in b:
    if i!=45:
        print("sudoku is not valid")
        break
    else:
        print("for rows it is valid")
```

```python
c=np.sum(s,axis=1)
for i in c:
    if i!=45:
        print("sudoku is not valid")
        break
    else:
        print("for colomns it is valid")
```
